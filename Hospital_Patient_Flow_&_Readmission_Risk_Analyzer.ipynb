```python
import sqlite3

conn = sqlite3.connect(':memory:')
cursor = conn.cursor()
```

```python
cursor.execute('''
CREATE TABLE patients (
    patient_id INTEGER,
    age INTEGER,
    condition TEXT,
    admission_type TEXT,
    length_of_stay INTEGER,
    readmitted TEXT,
    department TEXT
)
''')
```

```python
data = [
(1,65,'Diabetes','Emergency',5,'Yes','General'),
(2,45,'Cardiac','Planned',3,'No','Cardiology'),
(3,70,'Diabetes','Emergency',7,'Yes','General'),
(4,30,'Injury','Emergency',2,'No','Orthopedic'),
(5,55,'Cardiac','Planned',4,'Yes','Cardiology'),
(6,40,'Injury','Emergency',3,'No','Orthopedic'),
(7,68,'Diabetes','Emergency',6,'Yes','General'),
(8,50,'Cardiac','Planned',5,'No','Cardiology'),
(9,72,'Diabetes','Emergency',8,'Yes','General'),
(10,35,'Injury','Emergency',2,'No','Orthopedic')
]

cursor.executemany('INSERT INTO patients VALUES (?,?,?,?,?,?,?)', data)
conn.commit()
```

```python
import pandas as pd

query = '''
SELECT department, COUNT(*) as total_patients
FROM patients
GROUP BY department
'''

pd.read_sql(query, conn)
```

```python
query = '''
SELECT condition,
COUNT(*) as total,
SUM(CASE WHEN readmitted = 'Yes' THEN 1 ELSE 0 END) as readmitted_count
FROM patients
GROUP BY condition
'''

pd.read_sql(query, conn)
```

```python
query = '''
SELECT *
FROM patients
WHERE readmitted = 'Yes'
'''

pd.read_sql(query, conn)
```

```python
query = '''
SELECT readmitted, AVG(length_of_stay) as avg_stay
FROM patients
GROUP BY readmitted
'''

pd.read_sql(query, conn)
```

```python
query = '''
SELECT admission_type, COUNT(*) as total
FROM patients
GROUP BY admission_type
'''

pd.read_sql(query, conn)
```

Hospital Patient Flow and Readmission Risk Analyzer

Problem Statement  
Hospitals often face challenges in managing patient flow and reducing readmission rates. High readmission rates can indicate inefficiencies in treatment, discharge planning, or follow-up care. Understanding the factors contributing to readmissions is critical for improving patient outcomes and optimizing hospital resources.

Objective  
The objective of this project is to analyze hospital patient data to identify patterns in patient flow and determine key factors contributing to readmission risk. The analysis aims to provide actionable insights that can help improve operational efficiency and patient care.

Dataset Description  
The dataset used in this project is a simulated dataset representing hospital patient records. It includes the following attributes:

- Patient ID: Unique identifier for each patient  
- Age: Age of the patient  
- Condition: Primary medical condition  
- Admission Type: Emergency or planned admission  
- Length of Stay: Number of days admitted  
- Readmitted: Indicates whether the patient was readmitted  
- Department: Hospital department handling the patient  

This dataset is created to simulate real-world hospital scenarios where access to actual patient data is restricted.

Methodology  
The analysis was performed in the following steps:

1. Data creation and loading into the environment  
2. Data exploration to understand distributions and patterns  
3. Analysis of patient flow across departments and admission types  
4. Identification of readmission patterns using grouping and aggregation  
5. SQL-based analysis to query and extract key insights  
6. Visualization of trends to support findings  

Patient Flow Analysis  
This section analyzes patient distribution across departments and admission types to understand hospital workload and operational patterns.

Readmission Analysis  
This section focuses on identifying patterns in patient readmissions. The goal is to determine which factors such as condition, admission type, and length of stay contribute to higher readmission rates.

SQL Analysis  
SQL queries were used to perform structured analysis on the dataset. This includes aggregating patient counts, identifying high-risk groups, and evaluating trends in readmissions and hospital stays.

Key Insights  
The analysis reveals the following key insights:

- Patients with chronic conditions such as diabetes show higher readmission rates  
- Emergency admissions are more likely to result in readmissions compared to planned admissions  
- Longer hospital stays are associated with higher readmission probability  
- Certain departments handle a higher proportion of high-risk patients  

Recommendations  
Based on the insights, the following recommendations are proposed:

- Implement follow-up programs for high-risk patients after discharge  
- Improve discharge planning for patients with chronic conditions  
- Introduce a readmission risk scoring system  
- Allocate additional resources to departments with higher patient load  

Conclusion  
This project demonstrates how data analysis can be used to identify patterns in hospital operations and patient behavior. By combining Python-based analysis with SQL queries, meaningful insights were derived to support decision-making. The findings can help improve patient outcomes and optimize resource utilization in healthcare systems.
